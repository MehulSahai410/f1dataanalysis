```python
#firstly would like to convert the grid into numeric and replace \N with nan and then we drop it
res['grid']=pd.to_numeric(res['grid'].replace(r'\N','np.nan'),errors='coerce')
res=res.dropna(subset=['grid']).copy()
```

```python
res
```

```python
for col in res.columns:
    count = (res['grid'] == 0).sum()
    print(count)
```

```python
#grouping the grid on the basis of their raceId
race_maximum_grid=res.groupby('raceId')['grid'].transform('max')
```

```python
res['grid_clean'] = np.where(res['grid'] == 0, race_maximum_grid + 1, res['grid']).astype(int)
```

```python
res.head()
```

```python
print("res columns:", res.columns.tolist())
print("df columns:", df.columns.tolist())
```

```python
#we merged the driver.csv with res.csv
merged=res.merge(df[['driverId', 'forename', 'surname', 'code']], on='driverId', how='left')
merged['Driver name']=merged['forename']+' '+merged['surname']
```

```python
merged.head(5)
```

```python
merged=merged.drop(columns=['forename','surname'])
```

```python
merged
```

```python
#final technique
merged['Posgain/loss']=merged['grid_clean']-merged['positionOrder']
```

```python
merged.head(5)
```

```python
#What is distribution of starting grid position wrt final position?
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10, 5))
ax=sns.countplot(data=merged[merged['grid_clean'] <= 24], x='grid_clean', palette='viridis')
plt.title('Distribution of Starting Grid Positions')
plt.xlabel('Grid Position')
plt.ylabel('Count of Drivers')
for container in ax.containers:
    ax.bar_label(container, fontsize=9, padding=3)
plt.show()
```

```python
#2nd question
#What is the relation between starting grid with the final position
#Pearson is used for linear relationship whereas Spearman is used for monotonic relation
# Correlation check
corr_pearson = merged['grid_clean'].corr(merged['positionOrder'], method='pearson')
corr_spearman = merged['grid_clean'].corr(merged['positionOrder'], method='spearman')
print(f"Pearson Correlation: {corr_pearson:.3f}")
print(f"Spearman Rank Correlation: {corr_spearman:.3f}")
plt.figure(figsize=(8, 6))
sns.regplot(
    data=merged.sample(2500, random_state=30),
    x='grid_clean',
    y='positionOrder',
    scatter_kws={'alpha': 0.2},
    line_kws={'color': 'red'}
)
plt.title('Starting Grid vs Finishing Position')
plt.xlabel('Starting Grid Position')
plt.ylabel('Finishing Position')
plt.show()
#final analysis i understood ki jo P1-5 pe qualify kia toh uska uss position me end krne k chances zyada hai
```

```python
#3rd Question
#Which drivers gain the most positions on average during a race?
driver_stats = merged.groupby('Driver name').agg(
    races_started=('positionOrder', 'count'),
    avg_positions_gained=('Posgain/loss', 'mean')
).reset_index()

top_overtakers = driver_stats[driver_stats['races_started'] >= 40].sort_values(
    by='avg_positions_gained', ascending=False
).head(10)

print("Top 10 Drivers Gaining Most Positions on Average:")
print(top_overtakers)
```

```python
#4th question
#people with wins from pole positions
# Grid == 1 wale cases
pole_starts = merged[merged['grid_clean'] == 1]
pole_wins = pole_starts[pole_starts['positionOrder'] == 1]

pole_win_pct = (len(pole_wins) / len(pole_starts)) * 100
print(f"Total Pole Starts: {len(pole_starts)}")
print(f"Wins from Pole: {len(pole_wins)}")
print(f"Pole Position Win Percentage: {pole_win_pct:.2f}%")
```

```python
#top drivers best recovering from behind
qualify_low = merged[merged['grid_clean'] >= 11]
recovery_stats = qualify_low.groupby('Driver name').agg(
    poor_starts_count=('positionOrder', 'count'),
    avg_pos_gain=('Posgain/loss', 'mean'),
    points_finishes=('positionOrder', lambda x: (x <= 10).sum())
).reset_index()

best_recovery = recovery_stats[recovery_stats['poor_starts_count'] >= 25].sort_values(
    by='avg_pos_gain', ascending=False
).head(10)

print("Top 10 Drivers Best at Recovering from P11+:")
print(best_recovery)
```
